```python
**Variables & Data Types**


*A Variable is a container for storing data. You don't need to declare what type of data it is; Python is smart enough to figure it out!* ☁


▶ Core Data Types :


✅ String (str): Text wrapped in quotes (e.g., "Hello").

✅ Integer (int): Whole numbers (e.g., 10, -5).

✅ Float (float): Decimal numbers (e.g., 10.5, 3.14).

✅ Boolean (bool): True or False.
```

```python
# Creating variables
first_name = "Maneesha"   # String
age = 25                  # Integer
height = 1.75             # Float
is_student = False        # Boolean

# Printing the variables and checking their types
print("Name:", first_name, "| Type:", type(first_name))
print("Age:", age, "| Type:", type(age))
print("Height:", height, "| Type:", type(height))
print("Is Student?", is_student, "| Type:", type(is_student))
```
